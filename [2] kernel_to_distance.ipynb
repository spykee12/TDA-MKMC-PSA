Check if the CN filtered kernel matrices are positive semi-definite

```python
def is_positive_semi_definite(matrix):
    eigenvalues = np.linalg.eigvalsh(matrix)  # Compute eigenvalues
    return np.all(eigenvalues >= -1e-5), eigenvalues  # Check if all are non-negative

#matrix = pd.read_excel("linear_kernel_matrix2318_CN.xlsx", index_col=0) 
#matrix = pd.read_excel("rbf_kernel_matrix2318_CN.xlsx", index_col=0) 
matrix = pd.read_excel("mixed_kernel_matrix2318_CN.xlsx", index_col=0) 
#print(is_positive_semi_definite(matrix))  # Output: True or False
```

If not positive definite, do this:

```python
def make_positive_semi_definite(matrix, epsilon=1e-6):
    while not is_positive_semi_definite(matrix):
        matrix += epsilon * np.eye(matrix.shape[0])
        epsilon *= 10
    return matrix
```

Convert kernel matrix to distance matrix

```python
import numpy as np
import pandas as pd

def compute_kernel_distance(K):
    D_squared = 2 * (1 - K.values)  # Apply the formula
    D_squared[D_squared < 0] = 0  # Ensure non-negative values (numerical stability)
    D = np.sqrt(D_squared)  # Take the square root to get distances

    # Convert back to DataFrame
    return pd.DataFrame(D, index=K.index, columns=K.columns)
```

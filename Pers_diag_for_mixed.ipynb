```python
import pandas as pd
import numpy as np
import gudhi as gd
import matplotlib.pyplot as plt

# Step 1: Load the Excel file
file_path = 'mixed_distance_matrix_580v1_forPD.xlsx'
df = pd.read_excel(file_path, header=None)

# Step 2: Convert all entries to numeric, coercing errors to NaN
df = df.apply(pd.to_numeric, errors='coerce')

# Step 3: Check for NaN or invalid entries
if df.isnull().values.any():
    raise ValueError("The distance matrix contains non-numeric or missing values. Please clean your Excel file.")

# Step 4: Convert to NumPy array
distance_matrix = df.to_numpy()

# Step 5: Validate symmetry and zero diagonal
if not np.allclose(distance_matrix, distance_matrix.T, atol=1e-8):
    raise ValueError("Distance matrix is not symmetric.")
if not np.allclose(np.diag(distance_matrix), 0, atol=1e-4):
    raise ValueError("Distance matrix diagonal is not all zeros.")

# Step 6: Rips Complex and Persistence
rips_complex = gd.RipsComplex(distance_matrix=distance_matrix, max_edge_length=np.max(distance_matrix))
simplex_tree = rips_complex.create_simplex_tree(max_dimension=2)
simplex_tree.compute_persistence()
diag = simplex_tree.persistence()

# Step 7: Output
#print("Persistence Diagram:")
#for interval in diag:
    #print(interval)

gd.plot_persistence_diagram(diag)
plt.title("Persistence Diagram for Group 1")
plt.show()
```

```python
# Step 1: Retrieve Betti numbers across filtration values
barcodes = simplex_tree.persistence_intervals_in_dimension(0)
betti_0_births = barcodes[:, 0]
betti_0_deaths = barcodes[:, 1]

barcodes_1 = simplex_tree.persistence_intervals_in_dimension(1)
betti_1_births = barcodes_1[:, 0]
betti_1_deaths = barcodes_1[:, 1]

# Combine all filtration values for analysis
filtration_values = np.linspace(0, np.max(distance_matrix), 100)
betti_0_counts = []
betti_1_counts = []

for eps in filtration_values:
    b0 = np.sum((betti_0_births <= eps) & (betti_0_deaths > eps))
    b1 = np.sum((betti_1_births <= eps) & (betti_1_deaths > eps))
    betti_0_counts.append(b0)
    betti_1_counts.append(b1)

# Step 2: Plot Betti numbers
plt.figure(figsize=(10, 5))
plt.plot(filtration_values, betti_0_counts, label="Betti-0 (components)")
plt.plot(filtration_values, betti_1_counts, label="Betti-1 (holes)")
plt.xlabel("Filtration value")
plt.ylabel("Count")
plt.title("Betti Number Distribution")
plt.legend()
plt.grid(True)
plt.show()
```
